# Fine-tuning a Robot Policy

In this course you will take a real-robot vision-language-action model, MolmoAct2, and adapt it to a new skill with a short LoRA fine-tune, then drive the fine-tuned policy live in the LIBERO simulator. Everything runs in the browser on a single AMD Strix Halo machine, so inference and training remain local.

No workshop asset bundle is required. The notebooks download the pinned public checkpoints and datasets from Hugging Face on first use and reuse the persistent cache afterwards. The first run is large; see `README.md` for disk and network requirements and set `HF_TOKEN` when available.

## Overview

1. [Fine-tune MolmoAct2 on LIBERO](1_finetune_molmoact2_libero.ipynb) - load the public DROID base model and LIBERO dataset, sanity-check it open-loop, run a LoRA fine-tune, and evaluate the checkpoint you produced.
2. [Interactive simulation](2_interactive_sim_molmoact2_libero.ipynb) - command your checkpoint in natural language, or automatically use AllenAI's public trained LIBERO release.
3. [FastWAM world model](3_inference_fastwam_libero.ipynb) - download the public FastWAM release, drive LIBERO, and watch the model *imagine* the future video, not just the next actions.

## Outcomes

After you've completed this set of materials you will understand how LoRA adapts a large vision-language-action model cheaply enough to fine-tune on a single iGPU, the real-to-sim transfer story that makes the base a good starting point, and how to evaluate and drive a fine-tuned robot policy in a physics simulator. You will have run the whole loop - load, sanity check, fine-tune, evaluate - on AMD ROCm hardware, and you will know which knobs to turn to go from a workshop smoke-test to real training.
