# Local Robot Inference

In this tutorial you will learn how to serve language and vision models on your Radeon iGPU, drive a simulated robot arm with them over ROS 2, and have a model write and then optimize the robot's control code. Everything runs on this machine, so nothing you send leaves for a cloud endpoint.

## Overview

1. [Local inference](1_local_inference.ipynb) - first we will serve an LLM and a vision language model with Lemonade, and drive both through its OpenAI-compatible API.
2. [Robot agents](2_robot_agents.ipynb) - you will command a simulated Franka Panda arm in natural language with RAI, which binds ROS 2 perception and motion tools to that local model.
3. [Code as policy](3_code_as_policy.ipynb) - the model will write a whole Python policy up front, which CaP-X runs on the arm in Robosuite and scores.
4. [Robot harness optimization](4_robot_harness_optimization.ipynb) - we will read a recorded RHO run that evolved a policy repository using a coding agent, then watch the deployed policy run live.

## Outcomes

After you've completed this set of materials you will be familiar with serving models locally on AMD hardware, and with three ways a language model can reach a robot: inside the control loop as an agent, upstream of it as a program generator, and offline as a code optimizer. You will have learned how to point any OpenAI-compatible tool at a model running on your own machine. And you will walk away with ROS 2 and Robosuite pipelines to adapt to your own robotics projects.
