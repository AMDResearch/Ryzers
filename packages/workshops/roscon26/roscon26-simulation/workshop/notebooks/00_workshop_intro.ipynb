# ROSCon Humanoid Locomotion Workshop

Welcome to the workshop hosted by **AMD** and **Robotec.ai**, running on an **AMD Strix Halo
mini-PC**. We will use GPU-accelerated simulation and reinforcement learning to adapt a Unitree G1
humanoid policy from flat-ground walking to stair climbing.

## The two workshop notebooks

1. [Humanoid locomotion — training](humanoid_locomotion_train.ipynb) configures the stair terrain,
   rewards, curriculum, and PPO fine-tuning run. Start it first in its own Jupyter kernel and let it
   train in the background. It writes the result to
   `logs/roscon_stairs/model_stairs.pt`.
2. [Humanoid locomotion — evaluation](humanoid_locomotion_eval.ipynb) visualizes the generated
   terrain, measures the flat-ground baseline, plays its failure case, and then evaluates and plays
   the stair-trained policy.
